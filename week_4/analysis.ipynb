📌 Cell 1: Import Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

📌 Cell 2: Load the Dataset (Error Handling Included)

```python
try:
    df = pd.read_csv("D:\Developer Arena\week_4\sales_data.csv")
    print("Dataset loaded successfully!")
except FileNotFoundError:
    print("Error: File not found. Please check the file path.")
```

📌 Cell 4: Data Cleaning & Validation

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Handle missing values
df = df.fillna(method="ffill")

# Check for remaining missing values
df.isnull().sum()
```

📌 Cell 5: Save Cleaned Dataset

```python
df.to_csv("D:\Developer Arena\week_4\cleaned_sales_data.csv", index=False)
print("Cleaned dataset saved successfully.")
```

```python
df
```

📌 Cell 6: Basic Analysis & Metrics

```python
# Total Sales
total_sales = df["Total_Sales"].sum()

# Average Sales
average_sales = df["Total_Sales"].mean()

# Best Selling Product
best_product = df.groupby("Product")["Total_Sales"].sum().idxmax()

total_sales, average_sales, best_product
```

📌 Cell 7: Visualization 1 – Sales by Product (Bar Chart)

```python
plt.figure()
df.groupby("Product")["Total_Sales"].sum().plot(kind="bar")
plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total_Sales")
plt.show()
```

📌 Cell 8: Visualization 2 – Sales Trend Over Time (Line Chart)

```python
if "Date" in df.columns:
    df["Date"] = pd.to_datetime(df["Date"])
    plt.figure()
    df.groupby("Date")["Total_Sales"].sum().plot()
    plt.title("Sales Trend Over Time")
    plt.xlabel("Date")
    plt.ylabel("Sales")
    plt.show()
else:
    print("Date column not found in dataset.")
```

📌 Cell 9: Insights & Findings

```python
print("📌 Key Insights:")
print(f"• Total Sales: {total_sales}")
print(f"• Average Sales: {average_sales}")
print(f"• Best Selling Product: {best_product}")
```
